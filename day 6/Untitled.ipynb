```python
# 01 list comprehension masterclass
s=[45,82,91,33,67,55,78,29,95,61]
n=["rahul","priya","arjun","sneha","vikram"]
d=[i for i in s if i>60]
squre=[i**2 for i in s]
for i in s:
    if i>=80:
        print(i,"A")
    elif i>=60:
        print(i,"B")
    else:
        print(i,"C")
title=[i.title() for i in n]
ab=[i for i in s if i>50 and i%2==0]
print("passing:",d)
print("__________________________________")
print("Squre:",squre)
print("__________________________________")
print("name title case:",title)
print("__________________________________")
print("above 50 & even:",ab)
```

```python
# 02 FizzBuzz - the classic
def fizzbuzz(n):
    
    for i in range(1,n+1):
        if i%15==0:
            print("FizzBuzz")
        elif i%3==0:
            print("Fizz")
        elif i%5==0:
            print("Buzz")
        else:
            print(i)
fizzbuzz(20)
```

```python
#03 Caesar Cipher
def encr(t,s):
    r=""
    for i in t:
        if i.isupper():
            r=r+chr((ord(i)-ord("A")+s)%26+ord("A"))
        elif i.islower():
            r=r+chr((ord(i)-ord("a")+s)%26+ord("a"))
        else:
            r=r+i
    return r
def decr(t,s):
    return encr(t,26-s)
print(encr("Hello World",3))
print(decr("Khoor Zruog",3))
```

```python
# 04 number of word
on=["zero","one","tow","three","four","five","six","seven","eight","nine","ten","eleven",'twelve','thirteen','fourteen',
    "fifteen",'sixteen',"seventeen",'eighteen','nineteen']
te=["","","twenty","thirty","forty","fity",'sixty','seventy','eighty','ninety']
def show(n):
    if n==0:
        return on[n]
    elif n<20:
        return on[n]
    elif n<100:
        return te[n//10]+" "+on[n%10]
    elif n<1000:
        return on[n//100]+"hundred"+show(n%100)
    
s=[1,0,7,15,45,100,305,999]
for i in range(len(s)):
    print(s[i],show(s[i]))
```

```python
# 05 Bank account 
class ba:
    def __init__(self,owner,balance):
        self.owner=owner
        self.balance=balance
        self.his=[]
    def dep(self,amount):
        self.balance+=amount
        self.his.append(amount)
        return self.balance
    def withdraw(self,amount):
        if amount<0:
            print("invaliv amount")
        if amount>self.balance:
            print("insufficent balance")
        else:
            self.balance-=amount
            self.his.append(amount)
            
        return self.balance
        
    def stm(self):
        print("Owner name:",self.owner)
        print("acc balance:",self.balance)
        print("history:",self.his)
a=ba("rahul",1000)
a.dep(500)
a.withdraw(100)
a.stm()
print("_________________________")
a=ba("abhay",100000)
a.dep(5000)
a.withdraw(1000)
a.stm()
      
```
