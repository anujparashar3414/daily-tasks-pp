```python
def palindrome(n):
    n=str(n)
    if len(n)>1:
        if n==n[::-1]:
            return True
        else:
            return False
    else:
        return False
def armstrong(n):
    a=n
    n=str(n)
    if len(n)>1:
        l=len(n)
        n=int(n)
        s=0
        while n>0:
            r=n%10
            s=s+r**l
            n=n//10
        if a==s:
            return True
    else:
        return False
def prime(n):
    if n==0 or n==1:
        pass
    else:
        for i in range(2,n):
            if n%i==0:
                return False
        return True
def classify(n):
    l=[]
    if palindrome(n):
        l.append("palindrome")
    if armstrong(n):
        l.append("armstrong")
    if prime(n):
        l.append("prime")
    if l:
        return l
    else: 
        ["None"]
n=[2,11,121,153,370,1034]
for i in n:
    print(i,classify(i))
        
```

```python
cart=[
    {"item":"laptop","price":45000,"qty":1},
    {"item":"mouse","price":800,"qty":2},
    {"item":"keyboard","price":1500,"qty":1},
    {"item":"pen drive","price":600,"qty":3},
    {"item":"charger","price":1200,"qty":1},
]
total=0
after=[]
b=0
mulqty=[]
for i in cart:
    total+=i["price"]*i["qty"]
    after.extend([i["item"],i["price"]-(i['price']*10//100)])
    if i['price']>b:
        b=i["price"]
        c=i["item"]
    if i["qty"]>1:
        mulqty.append(i['item'])
        
print("total bill:",total)
print("after discount:",after)
print('most expensive:',c,b)
print("multy qty:",mulqty)
    
```

```python
class emp:
    def __init__(self,name,empid):
        self.name=name
        self.empid=empid
    def getpay(self):
        pass
    def display(self):
        print(self.empid,self.name,self.getpay())
    
class fulltime(emp):
    def __init__(self,name,empid,monthlysal):
        super().__init__(name,empid)
        self.monthlysal=monthlysal
    def getpay(self):
        return self.monthlysal
class partime(emp):
    def __init__(self,name,empid,hrate,hworked):
        super(). __init__ (name,empid)
        self.hrate=hrate
        self.hworked=hworked
    def getpay(self):
        return self.hrate*self.hworked
emps=[
    fulltime("rahul",'E01',45000),
    partime("priya",'E02',500,80),
    fulltime("arjun",'E03',45000),
    partime("sneha",'E04',600,65)
    ]
for i in emps:
    i.display()
        
        
        
        
```
