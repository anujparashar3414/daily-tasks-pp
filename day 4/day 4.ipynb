```python
def triangle(n):
    for i in range(1,n+1):
        for j in range(1,i+1):
            print("*",end="")
        print()

def squre(n):
    for i in range(1,n+1):
        for j in range(1,5):
            print("*",end="")
        print()
def reverse(n):
    for i in range(1,n+1):
        for j in range(1,6-i+1):
            print("*",end="")
        print()
triangle(5)
print("-------------")
squre(4)
print("-------------")
reverse(5)
```

```python
s="the cat sat on the mat the cat sat"
s=s.split(" ")
d={}
c={}
m=1
e=set()

for i in s:
    d.update({i:s.count(i)})
    if s.count(i)>m:
        m=s.count(i)
        c.update({i:m})
    if s.count(i)>1:
        e.add(i)
        
    

print("word",s)
print("frequency",d)
print("most frequency",c)
print("repeated words",e)
    
```
