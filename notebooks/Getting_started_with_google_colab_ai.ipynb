```python
from google.colab import drive
drive.mount("/content/drive")
import pandas as pd
df=pd.read_excel("/content/drive/MyDrive/1000.xlsx")
df.dropna(inplace=True)
import matplotlib.pyplot as plt
plt.subplot(1,2,1)
plt.plot(df["Discount"])
plt.subplot(2,2,2)
plt.scatter(df["Discount"],df["Unit_Price"])
plt.subplot(2,2,3)
plt.bar(df["Discount"],df["Unit_Price"])
plt.subplot(2,2,4)
plt.pie(df["Discount"],labels=df["Product"])
plt.show()
```

```python
def upp(col1,col2):
  return col1*col2
df["tot"]=df.apply(lambda x:upp(x["Quantity"],x["Unit_Price"]))
df
```

```python
import matplotlib.pyplot as plt
df.dropna(inplace=True)
plt.figure(figsize=(12,15))
plt.subplot(3,2,1)
plt.plot(df["tot"].head(50),marker="o",c="red",ls="dashed")
plt.subplot(3,2,2)
plt.scatter(df["tot"].head(50),df["Discount"].head(50),c="blue")
plt.subplot(3,2,3)
plt.bar(df["tot"],df["Discount"])
plt.subplot(3,2,4)
plt.pie(df["tot"].head(),labels=df["Discount"].head(),autopct="%1.1f%%")
plt.subplot(3,2,5)
plt.hist(df["tot"],bins=5)
plt.subplot(3,2,6)
plt.boxplot(df["tot"])
plt.show()
```
